## Some filtering/mutation operations with Pandas (groupbys)

```python
buildings['Bldg Status'].unique()
```

```python
buildings.groupby('Bldg Status')
```

```python
for status, group in buildings.groupby('Bldg Status'):
    print(status)
```

```python
for status, group in buildings.groupby('Bldg Status'):
    print(status)
    print(group)
```

```python
buildings.groupby('Bldg Status')['Square Footage'].sum() # what is the total square footage of buildings in each category
```

```python
buildings.groupby('Year Acquired')['Square Footage'].sum() # total square footage in each year acquired
```

```python
fig, ax = plt.subplots(figsize=(10,2))
gbs = buildings.groupby('Year Acquired')['Square Footage'].sum()
gbs.plot(ax=ax)
ax.set_yscale('log')
ax.set_ylabel('Square Footage')
plt.show() # supress output (and other stuff)
```

```python
x = gbs.index.values # useful for pulling the index out of series
```

```python
y = gbs.values # pulling the values out of a series
#y
```

```python
fig, ax = plt.subplots(figsize=(10,2))
gbs = buildings.groupby('Year Acquired')['Square Footage'].sum()
# gbs.plot(ax=ax)
ax.plot(x,y)
ax.set_yscale('log')
ax.set_ylabel('Square Footage')
ax.set_xlabel('Year Acquired')
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10,2))
gbs = buildings.groupby('Year Acquired')['Square Footage'].sum()
# gbs.plot(ax=ax)
ax.plot(x,y, marker='.')
ax.set_yscale('log')
ax.set_ylabel('Square Footage')
ax.set_xlabel('Year Acquired')
plt.show()
```

```python
stats = buildings.groupby('Year Acquired')['Square Footage'].describe()
stats
```

```python
fig, ax = plt.subplots(figsize=(10,2))
ax.plot(stats['max'], marker='.')
ax.set_yscale('log')
ax.set_ylabel('Square Footage')
ax.set_xlabel('Year Acquired')
plt.show()
```
