## A few items with bqplot (before moving on to other things)

There are more things with bqplot in prep & extra notebooks!

```python
import bqplot
```

```python
# 1. data
shape_data = bqplot.topo_load("map_data/WorldMap.json") # map_data folder?
```

```python
# 2. scales/axes
sc_geo = bqplot.Mercator() # know this is a bad projection!
```

```python
# 3. mark(s)
# bqplot.Map?
shape_map = bqplot.Map(map_data = shape_data, scales={'projection':sc_geo})
```

```python
# 4-5 could be interactivity

# 6. put all together as figure
fig = bqplot.Figure(marks=[shape_map], fig_margin={'top':0, 'bottom':0,'left':0, 'right':0})
fig
```

### United states map

```python
# 1. data
shape_data = bqplot.topo_load('map_data/USStatesMap.json') # map data? from where?

# 2. scales/axes
sc_geo = bqplot.Mercator() # know this is a bad projection!

# 3. mark(s)
shape_map = bqplot.Map(map_data = shape_data, scales={'projection':sc_geo})

# 6. put all together as figure
fig = bqplot.Figure(marks=[shape_map], fig_margin={'top':0, 'bottom':0,'left':0, 'right':0})
fig
```

```python
# 1. data
shape_data = bqplot.topo_load('map_data/USStatesMap.json') # map data? from where?

# 2. scales/axes
# sc_geo = bqplot.Mercator() # know this is a bad projection!
sc_geo = bqplot.AlbersUSA()

# 3. mark(s)
shape_map = bqplot.Map(map_data = shape_data, scales={'projection':sc_geo})

# 6. put all together as figure
fig = bqplot.Figure(marks=[shape_map], fig_margin={'top':0, 'bottom':0,'left':0, 'right':0})
fig
```

### Combining layers of geospatial data

```python
import matplotlib.pyplot as plt
```

```python
fig, ax = plt.subplots(figsize=(4,4))
gdf_councils.plot(ax=ax)
gdf_apts.plot(ax=ax, color='lightgreen')
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(4,4))
gdf_councils.plot(ax=ax)
gdf_apts.plot(ax=ax, color='lightgreen')
ax.set_xlim(-88.245, -88.225)
ax.set_ylim(40.10, 40.125)
plt.show()
```

```python
gdf_apts['Building_Type'].unique()
```

```python
fig, ax = plt.subplots(figsize=(4,4))
gdf_councils.plot(ax=ax)
#gdf_apts.plot(ax=ax, color='lightgreen')
gdf_apts.plot(ax=ax, column='Building_Type', cmap='rainbow', legend=True)
ax.set_xlim(-88.245, -88.225)
ax.set_ylim(40.10, 40.125)
plt.show()
```
