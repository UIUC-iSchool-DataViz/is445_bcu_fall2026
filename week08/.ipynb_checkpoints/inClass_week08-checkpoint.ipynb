## Dashboards in Altair (through vega-lite)

````python
import pandas as pd
````

````python
df = pd.read_csv('https://raw.githubusercontent.com/UIUC-iSchool-DataViz/is445_data/main/mobility.csv')
````

````python
df.head()
````

````python
df.columns
````

What we want is a dashboard:
1. on the left/driver side (drives changes to right plot) we want a heatmap of State (y-axis) and Student-to-teacher-ratio (x-axis)
2. On responder side (right plot) we want histogram of Mobility score

Idea: select region of State and Student-to-teacher-ratio and see updates to Mobility score

First, quick python-only prototype:

````python
df['Mobility'].plot(kind='hist', figsize=(2,2))
````

````python
df['Student_teacher_ratio'].plot(kind='hist', figsize=(2,2), bins=10)
````

````python
import numpy as np
````

````python
bins = np.linspace(df['Student_teacher_ratio'].min(), df['Student_teacher_ratio'].max(), 10)
bins
````

````python
pd.cut(df['Student_teacher_ratio'],bins)
````

````python
table = df.pivot_table(index='State', columns=pd.cut(df['Student_teacher_ratio'],bins), aggfunc='size')
````

````python
table
````

````python
import matplotlib.pyplot as plt
````

````python
fig, ax = plt.subplots(figsize=(10,8))
# extent = (xmin, xmax, ymin, ymax)
extent = [bins.min(), bins.max(), 0, len(table.index)]
ax.imshow(table.values, cmap='hot', extent=extent)
ax.set_yticks(range(len(table.index)))
ax.set_yticklabels(table.index)
plt.show()
````

````python
chart2 = alt.Chart.from_dict(
{
    # Data
  "data": {"url": "https://raw.githubusercontent.com/UIUC-iSchool-DataViz/is445_data/main/mobility.csv"},
    # Mark
  "mark": "bar",
    # Axis/Scales
  "encoding": {
    "x": {
      #"bin": true,
      "bin": True,
      "field": "Mobility", 
      "axis":{"title":"Mobility Score"}
    },
    "y": {"aggregate": "count", "axis":{"title":"Frequency"}}
  }
}  
)
chart2
````

```javascript
{
  "$schema": "https://vega.github.io/schema/vega-lite/v6.json",
  "data": {"url": "data/cars.json"},
  "mark": "rect",
  "encoding": {
    "y": {"field": "Origin", "type": "nominal"},
    "x": {"field": "Cylinders", "type": "ordinal"},
    "color": {"aggregate": "mean", "field": "Horsepower"}
  },
  "config": {
    "axis": {"grid": true, "tickBand": "extent"}
  }
}
```

Not exactly what we want -- y-field is State (ordinal is fine), but x is binned (Student-teacher-ratio), and color should be counts, not another column (Horsepower)

````python
chart1 = alt.Chart.from_dict(
{
    # Data
  "data": {"url": "https://raw.githubusercontent.com/UIUC-iSchool-DataViz/is445_data/main/mobility.csv"},
    # Mark
  "mark": "rect",
    # height/width
    "height":400,
    # Encodings = scales + axis
  "encoding": {
    "x":{"field":"Student_teacher_ratio", "type":"quantitative", "bin":{"maxbins":10}},
    "y": {"field": "State", "type": "ordinal"},
    "color":{"aggregate":"count", "type":"ordinal"}
    #"x": {"field": "Cylinders", "type": "ordinal"},
    #"color": {"aggregate": "mean", "field": "Horsepower"}
  }
} 
)
chart1
````

````python
chart = alt.HConcatChart(hconcat=[chart1,chart2])
chart
````

````python
alt.selection_interval?
````

````python
brush = alt.selection_interval()
````

````python
chart1.add_params(brush)
````

````python
brush = alt.selection_interval(encodings=['x','y'])
````

````python
chart1 = alt.Chart.from_dict(
{
    # Data
  "data": {"url": "https://raw.githubusercontent.com/UIUC-iSchool-DataViz/is445_data/main/mobility.csv"},
    # Mark
  "mark": "rect",
    # height/width
    "height":400,
    # Encodings = scales + axis
  "encoding": {
    "x":{"field":"Student_teacher_ratio", "type":"quantitative", "bin":{"maxbins":10}},
    "y": {"field": "State", "type": "ordinal"},
    "color":{"aggregate":"count", "type":"ordinal"}
  }
} 
).add_params(
    brush
)
# chart1
````

````python
chart2 = alt.Chart.from_dict(
{
    # Data
  "data": {"url": "https://raw.githubusercontent.com/UIUC-iSchool-DataViz/is445_data/main/mobility.csv"},
    # Mark
  "mark": "bar",
    # Axis/Scales
  "encoding": {
    "x": {
      #"bin": true,
      "bin": True,
      "field": "Mobility", 
      "axis":{"title":"Mobility Score"}
    },
    "y": {"aggregate": "count", "axis":{"title":"Frequency"}}
  }
}  
).transform_filter(
    brush
)
# chart2
````

````python
chart = chart1 | chart2
````

````python
chart
````

## A bit more intuition with Altair

````python
mobility_url = 'https://raw.githubusercontent.com/UIUC-iSchool-DataViz/is445_data/main/mobility.csv'
````

````python
scatter1 = alt.Chart(mobility_url).mark_point().encode(
    x='Mobility:Q',
    y='Population:Q'
)
scatter1
````

````python
scatter2 = alt.Chart(mobility_url).mark_point().encode(
    x='Mobility:Q',
    y='Population:Q', 
    #color="Income:Q"
    color=alt.Color("Income:Q", scale=alt.Scale(scheme='sinebow'), bin=alt.Bin(maxbins=5))
)
scatter2
````

````python
scatter3 = alt.Chart(mobility_url).mark_point().encode(
    x='Mobility:Q',
    y=alt.Y('Population:Q', scale=alt.Scale(type='log')),
    color=alt.Color("Income:Q", scale=alt.Scale(scheme='sinebow'), bin=alt.Bin(maxbins=5))
)
scatter3
````

## Remake our dashboard with Altair (using mobility_url -- data stored on web)

```python
chart1 = alt.Chart.from_dict(
{
    # Data
  "data": {"url": "https://raw.githubusercontent.com/UIUC-iSchool-DataViz/is445_data/main/mobility.csv"},
    # Mark
  "mark": "rect",
    # height/width
    "height":400,
    # Encodings = scales + axis
  "encoding": {
    "x":{"field":"Student_teacher_ratio", "type":"quantitative", "bin":{"maxbins":10}},
    "y": {"field": "State", "type": "ordinal"},
    "color":{"aggregate":"count", "type":"ordinal"}
  }
} 
).add_params(
    brush
)
```

````python
chart1 = alt.Chart(mobility_url).mark_rect().encode(
    alt.X("Student_teacher_ratio:Q", bin=alt.Bin(maxbins=10)),
    alt.Y("State:O"),
    alt.Color("count()")
).properties(
    height=400
)
chart1
````
