```python
# import our usual things
import matplotlib.pyplot as plt
import matplotlib
import numpy as np
import pandas as pd
```

```python
buildings = pd.read_csv("https://github.com/UIUC-iSchool-DataViz/is445_data/raw/main/building_inventory.csv", 
                       na_values = {
                           'Square Footage':0,
                           'Year Acquired':0,
                           'Year Constructed':0,
                           'Total Floors':0
                       })
```

```python
#buildings.columns
```

```python
stats = buildings.groupby('Year Acquired')['Square Footage'].describe()
#stats
```

```python
fig, ax = plt.subplots(figsize=(10,2))
ax.plot(stats['max'], marker='.', linewidth=2.0)
ax.set_xlabel('Year Acquired')
ax.set_ylabel('Square Footage')
ax.set_yscale('log')
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10,2))
ax.plot(stats['max'], marker='.', linewidth=2.0, label='Max')
ax.plot(stats['min'], marker='.', linewidth=2.0, label='Min')
ax.set_xlabel('Year Acquired')
ax.set_ylabel('Square Footage')
ax.set_yscale('log')
ax.legend()
plt.show()
```

```python
with plt.style.context('_classic_test_patch'):
    # selected everything and then did CMD+]
    fig, ax = plt.subplots(figsize=(10,2))
    ax.plot(stats['max'], marker='.', linewidth=2.0, label='Max')
    ax.plot(stats['min'], marker='.', linewidth=2.0, label='Min')
    ax.set_xlabel('Year Acquired')
    ax.set_ylabel('Square Footage')
    ax.set_yscale('log')
    ax.legend()
    plt.show()
```

```python
# Step 1: hard-code the option that is updating style (but whatever 
# we want to make interactive -- set this as a hard-coded variable
plotting_style = '_classic_test_patch'
with plt.style.context(plotting_style):
    fig, ax = plt.subplots(figsize=(10,2))
    ax.plot(stats['max'], marker='.', linewidth=2.0, label='Max')
    ax.plot(stats['min'], marker='.', linewidth=2.0, label='Min')
    ax.set_xlabel('Year Acquired')
    ax.set_ylabel('Square Footage')
    ax.set_yscale('log')
    ax.legend()
    plt.show()
```

```python
# Step 2: have this parameter as an input to a function
def make_plot(plotting_style):
    # plotting_style = '_classic_test_patch'
    with plt.style.context(plotting_style):
        fig, ax = plt.subplots(figsize=(10,2))
        ax.plot(stats['max'], marker='.', linewidth=2.0, label='Max')
        ax.plot(stats['min'], marker='.', linewidth=2.0, label='Min')
        ax.set_xlabel('Year Acquired')
        ax.set_ylabel('Square Footage')
        ax.set_yscale('log')
        ax.legend()
        plt.show()
```

```python
# step 2.5: try out the function!
make_plot('seaborn')
```

```python
# step 3: decorate the function with an ipywidgets.interact with 
# the variable as an input to function and parameter in .interact
@ipywidgets.interact(plotting_style=plt.style.available)
def make_plot(plotting_style):
    with plt.style.context(plotting_style):
        fig, ax = plt.subplots(figsize=(10,2))
        ax.plot(stats['max'], marker='.', linewidth=2.0, label='Max')
        ax.plot(stats['min'], marker='.', linewidth=2.0, label='Min')
        ax.set_xlabel('Year Acquired')
        ax.set_ylabel('Square Footage')
        ax.set_yscale('log')
        ax.legend()
        plt.show()
```

```python
plt.imshow(mich, cmap='Reds')
plt.colorbar()
plt.clim(-355,355) # easy way to give us a 0 at the center of colormap
plt.show()
```

```python
# Step 1: hard-code a parameter I want to make interactive
colormap = 'Reds'
plt.imshow(mich, cmap=colormap)
plt.colorbar()
plt.clim(-355,355) # easy way to give us a 0 at the center of colormap
plt.show()
```

```python
# Step 2: make a function with the parameter of interest as input
#colormap = 'Reds'
def make_mich_plot(colormap):
    plt.imshow(mich, cmap=colormap)
    plt.colorbar()
    plt.clim(-355,355) # easy way to give us a 0 at the center of colormap
    plt.show()
```

```python
# Step 2.5: testing the function
make_mich_plot('Greens')
```

```python
# plt.colormaps()
```

```python
# Step 3: decorate the function
# make parameter with variable we want interactive
@ipywidgets.interact(colormap = plt.colormaps())
def make_mich_plot(colormap):
    plt.imshow(mich, cmap=colormap)
    plt.colorbar()
    plt.clim(-355,355) # easy way to give us a 0 at the center of colormap
    plt.show()
```

```python
# Step 1: hard code variables of interest
colormap = 'terrain'
take_log = True

if take_log:
    # apply a symmetric log, base 10 to colormap
    norm = matplotlib.colors.SymLogNorm(10)
else:
    norm = matplotlib.colors.Normalize() # default, linear scale
    
plt.imshow(mich, cmap=colormap, norm=norm)
plt.colorbar()
plt.clim(-355,355) # easy way to give us a 0 at the center of colormap
plt.show()
```

```python
# Step 2: make this a function
# colormap = 'terrain'
# take_log = True

def make_mich_plot2(colormap='Reds', take_log=True):
    if take_log:
        # apply a symmetric log, base 10 to colormap
        norm = matplotlib.colors.SymLogNorm(10)
    else:
        norm = matplotlib.colors.Normalize() # default, linear scale
        
    plt.imshow(mich, cmap=colormap, norm=norm)
    plt.colorbar()
    plt.clim(-355,355) # easy way to give us a 0 at the center of colormap
    plt.show()
```

```python
# Step 2.5: always test our function!
make_mich_plot2(colormap='magma', take_log=False)
```

```python
#Step 3: decorate the function
@ipywidgets.interact(colormap=plt.colormaps(), take_log=True)
def make_mich_plot2(colormap='Reds', take_log=True):
    if take_log:
        # apply a symmetric log, base 10 to colormap
        norm = matplotlib.colors.SymLogNorm(10)
    else:
        norm = matplotlib.colors.Normalize() # default, linear scale
        
    plt.imshow(mich, cmap=colormap, norm=norm)
    plt.colorbar()
    plt.clim(-355,355) # easy way to give us a 0 at the center of colormap
    plt.show()
```
