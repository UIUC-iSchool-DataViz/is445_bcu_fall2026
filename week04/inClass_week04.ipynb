```python
# import our usual things
import matplotlib.pyplot as plt
import matplotlib
import numpy as np
import pandas as pd
```

```python
buildings = pd.read_csv("https://github.com/UIUC-iSchool-DataViz/is445_data/raw/main/building_inventory.csv", 
                       na_values = {
                           'Square Footage':0,
                           'Year Acquired':0,
                           'Year Constructed':0,
                           'Total Floors':0
                       })
```

```python
#buildings.columns
```

```python
stats = buildings.groupby('Year Acquired')['Square Footage'].describe()
#stats
```

```python
fig, ax = plt.subplots(figsize=(10,2))
ax.plot(stats['max'], marker='.', linewidth=2.0)
ax.set_xlabel('Year Acquired')
ax.set_ylabel('Square Footage')
ax.set_yscale('log')
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10,2))
ax.plot(stats['max'], marker='.', linewidth=2.0, label='Max')
ax.plot(stats['min'], marker='.', linewidth=2.0, label='Min')
ax.set_xlabel('Year Acquired')
ax.set_ylabel('Square Footage')
ax.set_yscale('log')
ax.legend()
plt.show()
```

```python
# Step 1: hard-code the option that is updating style (but whatever 
# we want to make interactive -- set this as a hard-coded variable
plotting_style = '_classic_test_patch'
with plt.style.context(plotting_style):
    fig, ax = plt.subplots(figsize=(10,2))
    ax.plot(stats['max'], marker='.', linewidth=2.0, label='Max')
    ax.plot(stats['min'], marker='.', linewidth=2.0, label='Min')
    ax.set_xlabel('Year Acquired')
    ax.set_ylabel('Square Footage')
    ax.set_yscale('log')
    ax.legend()
    plt.show()
```

```python
# Step 2: have this parameter as an input to a function
def make_plot(plotting_style):
    # plotting_style = '_classic_test_patch'
    with plt.style.context(plotting_style):
        fig, ax = plt.subplots(figsize=(10,2))
        ax.plot(stats['max'], marker='.', linewidth=2.0, label='Max')
        ax.plot(stats['min'], marker='.', linewidth=2.0, label='Min')
        ax.set_xlabel('Year Acquired')
        ax.set_ylabel('Square Footage')
        ax.set_yscale('log')
        ax.legend()
        plt.show()
```

```python
# step 3: decorate the function with an ipywidgets.interact with 
# the variable as an input to function and parameter in .interact
@ipywidgets.interact(plotting_style=plt.style.available)
def make_plot(plotting_style):
    with plt.style.context(plotting_style):
        fig, ax = plt.subplots(figsize=(10,2))
        ax.plot(stats['max'], marker='.', linewidth=2.0, label='Max')
        ax.plot(stats['min'], marker='.', linewidth=2.0, label='Min')
        ax.set_xlabel('Year Acquired')
        ax.set_ylabel('Square Footage')
        ax.set_yscale('log')
        ax.legend()
        plt.show()
```
