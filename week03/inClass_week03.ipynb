```python
im_data = np.array(im)
```

```python
im_data.shape # 483 rows, 430 columns, 4 color channels -- R=red, G=green, B=blue, A=alpha/opacity
```

```python
fig, ax = plt.subplots(figsize=(10,2))
ax.hist(im_data[:,:,0]) # [:,:,0] = all pixels in all rows and columns in Red color channel
plt.show()
```

```python
im_data[:,:,0].shape
```

```python
fig, ax = plt.subplots(figsize=(10,2))
ax.hist(im_data[:,:,0].flatten()) # [:,:,0] = all pixels in all rows and columns in Red color channel
plt.show()
```

```python
im_data[:,:,0].flatten().shape
```

```python
np.unique(im_data[:,:,0])
```

```python
fig, ax = plt.subplots(figsize=(10,2))
ax.hist(im_data[:,:,0].flatten(), bins=256) # [:,:,0] = all pixels in all rows and columns in Red color channel
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10,2))
ax.hist(im_data[:,:,0].flatten(), bins=256, color='red') # [:,:,0] = all pixels in all rows and columns in Red color channel
ax.set_xlabel('Red channel bins')
ax.set_ylabel('# of pixels per bin')
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10,2))
ax.hist(im_data[:,:,0].flatten(), bins=256, color='red') # [:,:,0] = all pixels in all rows and columns in Red color channel
ax.hist(im_data[:,:,1].flatten(), bins=256, color='green')
ax.set_xlabel('Red channel bins')
ax.set_ylabel('# of pixels per bin')
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10,2))
ax.hist(im_data[:,:,0].flatten(), bins=256, color='red') # [:,:,0] = all pixels in all rows and columns in Red color channel
ax.hist(im_data[:,:,1].flatten(), bins=256, color='green')
ax.hist(im_data[:,:,2].flatten(), bins=256, color='blue')
ax.hist(im_data[:,:,3].flatten(), bins=256, color='gray') # A, alpha, opacity = set as gray
ax.set_xlabel('Red channel bins')
ax.set_ylabel('# of pixels per bin')
plt.show()
```

```python
fig, ax = plt.subplots(4, 1, figsize=(10,6))
ax[0].hist(im_data[:,:,0].flatten(), bins=256, color='red') # [:,:,0] = all pixels in all rows and columns in Red color channel
ax[1].hist(im_data[:,:,1].flatten(), bins=256, color='green')
ax[2].hist(im_data[:,:,2].flatten(), bins=256, color='blue')
ax[3].hist(im_data[:,:,3].flatten(), bins=256, color='gray') # A, alpha, opacity = set as gray
#ax.set_xlabel('Red channel bins')
#ax.set_ylabel('# of pixels per bin')
plt.show()
```

```python
fig, ax = plt.subplots(4, 1, figsize=(10,6))
ax[0].hist(im_data[:,:,0].flatten(), bins=256, color='red') # [:,:,0] = all pixels in all rows and columns in Red color channel
ax[1].hist(im_data[:,:,1].flatten(), bins=256, color='green')
ax[2].hist(im_data[:,:,2].flatten(), bins=256, color='blue')
ax[3].hist(im_data[:,:,3].flatten(), bins=256, color='gray') # A, alpha, opacity = set as gray
ax[0].set_xlabel('Red channel bins')
ax[1].set_xlabel('Green channel bins')
ax[2].set_xlabel('Blue channel bins')
ax[3].set_xlabel('Opacity channel bins')
#ax.set_ylabel('# of pixels per bin')
fig.tight_layout() # fits all plots in tightest layout without overlaps
plt.show()
```
