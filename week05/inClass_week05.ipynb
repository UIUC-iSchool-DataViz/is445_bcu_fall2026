```python
import traitlets
```

```python
# class MyObject():
#     name = 'unnamed'
#     age = 0

class MyTraitletObject(traitlets.HasTraits):
    name = traitlets.Unicode('unnamed') # trailet-specific way of saying "string"
    age = traitlets.Int(0) # traitlets specific way of saying "int"
```

```python
my_t_obj = MyTraitletObject()
```

```python
my_t_obj.name, my_t_obj.age
```

```python
my_t_obj.name = 'Bob'
my_t_obj.age = 47
```

```python
my_t_obj.name, my_t_obj.age
```

```python
# 1. Read in (make) data
x = np.arange(100)
y = np.random.random(100)

# 2. Scales
x_sc = bqplot.LinearScale()
y_sc = bqplot.LinearScale()

# 3. Axis
ax_x = bqplot.Axis(scale = x_sc, label='X Value')
ax_y = bqplot.Axis(scale = y_sc, label = 'Y Value', orientation='vertical')

# 4. Mark
lines = bqplot.Lines(x=x,y=y, scales = {'x':x_sc,'y':y_sc})

# 5. interactivity would be added
pz = bqplot.interacts.PanZoom(scales={'x':[x_sc], 'y':[y_sc]})

# 6. Put it all together in a figure
fig = bqplot.Figure(marks=[lines], axes = [ax_x, ax_y], interaction=pz)
fig
```

```python
# 1. make/collect data
x = np.random.random(100)
y = np.random.random(100)

# 2. scales
x_sc = bqplot.LinearScale()
y_sc = bqplot.LinearScale()

# 3. Axes
ax_x = bqplot.Axis(scale=x_sc, label='X')
ax_y = bqplot.Axis(scale=y_sc, label='Y', orientation='vertical')

# 4. marks
scatters = bqplot.Scatter(x=x,y=y, scales={'x':x_sc, 'y':y_sc})

# 5. interactivity - select an interval on x-axis
selector = bqplot.interacts.FastIntervalSelector(scale=x_sc, marks=[scatters])

# 6. put it all together in figure
fig = bqplot.Figure(marks=[scatters], axes=[ax_x,ax_y], interaction=selector)
fig
```

```python
# 1. data
data = np.random.random((10,10))

# 2. scales
col_sc = bqplot.ColorScale(scheme='Greens')
y_sc = bqplot.OrdinalScale()
x_sc = bqplot.OrdinalScale()

# 3. axis
col_ax = bqplot.ColorAxis(scale=col_sc, orientation='vertical', side='right')
ax_x = bqplot.Axis(scale=x_sc, label='Column Bin')
ax_y = bqplot.Axis(scale=y_sc, label='Row Bin', orientation='vertical')

# 4. marks
heat_map = bqplot.GridHeatMap(color=data, scales={'color':col_sc, 'row':y_sc, 'column':x_sc}, 
                             interactions={'click':'select'}, 
                             anchor_style={'fill':'magenta'})

out = ipywidgets.Output()
# 5. interactivity -- add in a widget that shows what index is selected
def on_selected(change):
    with out:
       #print(change)
       #print(change['owner'].selected)
       if len(change['owner'].selected) == 1: # no multi-selection
           print(change['owner'].selected[0])

heat_map.observe(on_selected, 'selected')

# 6. figure all together
fig = bqplot.Figure(marks=[heat_map], axes=[col_ax, ax_x, ax_y])
#fig
ipywidgets.VBox([fig, out])
```
