```python
import traitlets
```

```python
# class MyObject():
#     name = 'unnamed'
#     age = 0

class MyTraitletObject(traitlets.HasTraits):
    name = traitlets.Unicode('unnamed') # trailet-specific way of saying "string"
    age = traitlets.Int(0) # traitlets specific way of saying "int"
```

```python
my_t_obj = MyTraitletObject()
```

```python
my_t_obj.name, my_t_obj.age
```

```python
my_t_obj.name = 'Bob'
my_t_obj.age = 47
```

```python
my_t_obj.name, my_t_obj.age
```

```python
def name_change(change): # event for buttons? change?
    print(change)

my_t_obj.observe(name_change, ['name'])
```

```python
my_t_obj.name = 'Linda'
```

```python
def name_change2(change):
    print("Hello, my name is", change['new'], 'and I am', change['owner'].age, 'years old.')

my_t_obj.observe(name_change2, ['name'])
```

```python
my_t_obj.name = 'Bob'
```

```python
my_t_obj.unobserve(name_change,['name'])
```

```python
my_t_obj.name = 'Linda'
```

```python
my_t_obj = MyTraitletObject() # the "lazy" way to only observe what you want

def name_change2(change):
    print("Hello, my name is", change['new'], 'and I am', change['owner'].age, 'years old.')

my_t_obj.observe(name_change2, ['name'])
```

```python
my_t_obj.name = 'Linda'
```

```python
def age_change(change):
    print("Hello, my name is", change['owner'].name, 'and I am', change['new'], 'years old.')

my_t_obj.observe(age_change, ['age'])
```
